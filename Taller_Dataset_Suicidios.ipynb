```python
#1-
#Realice la limpieza y transformación de las siguientes columnas:
#a. La columna ID_ANO que esta de tipo float64 cambiarla a tipo entero (int64)

df['ID_ANO'] = df['ID_ANO'].astype('str')
df['ID_ANO'] = df['ID_ANO'].str.replace('.', '', regex=False)
df['ID_ANO'] = df['ID_ANO'].replace('202', '2020')
#df.dtypes
df['ID_ANO'] = df['ID_ANO'].astype('int64')
df
```

```python
#renombrar columna
df = df.rename(columns={'FECHA_DE_ NOTIFICACION_ DEL_ EVENTO': 'FECHA_NOTIFICACION'})
# Elimina espacios en blanco accidentales al inicio y final de cada nombre de columna
df.columns = df.columns.str.strip()
df
```

```python
#Renombra todas las columnas para quitar los espacios en blanco accidentales en el nombre de la columna
# Renombra las columnas para facilitar su manipulación en el código
df = df.rename(
    columns={
        "ID_ANO": "ID_ANIO",
        "FECHA_DE_ NOTIFICACION_ DEL_ EVENTO": "FECHA_DE_NOTIFICACION_DEL_EVENTO",
        "UNIDAD_DE_MEDIDA_DE _LA_VICTIMA": "UNIDAD_DE_MEDIDA_DE_LA_VICTIMA",
        "AREA_ DE_ RESIDENCIA": "AREA_DE_RESIDENCIA",
        "BARRIO_ DE_ RESIDENCIA": "BARRIO_DE_RESIDENCIA",
        "SEGURIDAD_ SOCIAL": "SEGURIDAD_SOCIAL",
        "POBLACION_ A _CARGO_ ICBF": "POBLACION_A_CARGO_ICBF",
        "FECHA_ DEL_ HECHO": "FECHA_DEL_HECHO",
        "INTENTOS_ PREVIOS": "INTENTOS_PREVIOS",
        "NUMERO_ DE_ INTENTOS": "NUMERO_DE_INTENTOS",
        "CONFLICTO_ CON_ PAREJA_ O _EX_ PAREJA": "CONFLICTO_CON_PAREJA_O_EX_PAREJA",
        "ENFERMEDAD_ CRONICA_ DOLOROSA_ O_ DISCAPACITANTE": "ENFERMEDAD_CRONICA_DOLOROSA_O_DISCAPACITANTE",
        "PROBLEMAS_ ECONOMICOS": "PROBLEMAS_ECONOMICOS",
    }
)
```

```python
#1-
#b. La columna FECHA_DE_NOTIFICACION_DEL_EVENTO que está en object convertir en tipo fecha.

#df.dtypes
# Para tener en cuenta, el formato de la fecha debe coincidir en espacios y en orden,
# es decir, anio, mes y día como esta en el dataset.
df['FECHA_NOTIFICACION'] = pd.to_datetime(df['FECHA_NOTIFICACION'], format='%Y %b %d %H:%M:%S %p')

df.dtypes
```

```python
#1-
#c. Para todas las columnas donde el dato es: “SI/NO”. Transformar el tipo de dato
# "SI" por el dígito binario 1 y el dato "NO" por el dígito binario 0.
#df["mi_columna"] = (df["mi_columna"] == "SI").astype(int)
# Filtra las columnas donde los valores únicos (convertidos a mayúsculas) coincidan con 'SI' y 'NO'
cols_sino = [
    col for col in df.columns
    if set(df[col].dropna().astype(str).str.strip().str.upper().unique()).issubset({"SI", "NO"})
]
print("Columnas que contienen solo SI/NO:", cols_sino)

# Reemplaza SI por 1 y NO por 0 solo en las columnas identificadas
df[cols_sino] = df[cols_sino].replace({"SI": 1, "NO": 0, "si": 1, "no": 0})
df
```

🌵
df[col].dropna(): Ignora los valores nulos (NaN) para no interferir con la búsqueda.

🌵
.str.strip().str.upper(): Limpia espacios accidentales y convierte a mayúsculas para asegurar que detecte variantes como "si", "Si", "NO ".

🌵
.unique(): Extrae los valores distintos que existen en esa columna.

🌵
.issubset({"SI", "NO"}): Verifica si todos los valores de esa columna pertenecen al conjunto {"SI", "NO"} (útil incluso si una columna solo tiene "SI" o solo "NO" en sus registros).

```python
#1-
#d. La columna UNIDAD_DE_MEDIDA_DE _LA_VICTIMA cambie su valor por la palabra “AÑOS”.
df['UNIDAD_DE_MEDIDA_DE_LA_VICTIMA'] = df['UNIDAD_DE_MEDIDA_DE_LA_VICTIMA'].replace('AÃOS', 'AÑOS')
df['UNIDAD_DE_MEDIDA_DE_LA_VICTIMA'] = df['UNIDAD_DE_MEDIDA_DE_LA_VICTIMA'].replace('ANIOS', 'AÑOS')
df
```

✅**Pandas Group Data**
Una operación groupby implica una combinación de división del objeto, aplicación de una función y combinación de los resultados. Esto puede utilizarse para agrupar grandes cantidades de datos y calcular operaciones sobre estos grupos.
![](https://drive.google.com/uc?export=view&id=13RBVHMLZhE1U-D_54-9R_mRcMfVjLUgJ)

```python
# Agrupar por la columna "SEXO" y contar los registros
#✅
#Opción 2: Con groupby() (Ideal para seguir procesando datos)

#Si necesitas usar explícitamente el agrupamiento para obtener un DataFrame con el conteo:
cantidad_sexo = df.groupby("SEXO").size().reset_index(name="CANTIDAD")

print(cantidad_sexo)
```

```python
#2- Realice las siguientes consultas:
#e. Cuántas víctimas fueron remitidas a psicología

cantidad_victimas_remitidas_psicologia = df.groupby("REMITIDO_ A_ PSICOLOGIA").size().reset_index(name="CANTIDAD")
print(cantidad_victimas_remitidas_psicologia)
```
